```python

df_customer = spark.table('samples.bakehouse.sales_transactions')
display(df_customer)
```

Save it to my default catalog

```python
%sql
CREATE SCHEMA bake_data.bronze;

CREATE SCHEMA bake_data.silver;

CREATE SCHEMA bake_data.gold;
```

```python
df_customer.write \
  .mode("overwrite") \
  .format("delta") \
  .saveAsTable("bake_data.bronze.sales_transactions")
```
