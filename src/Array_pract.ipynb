```python
df = spark.createDataFrame(data, schema=schema)
df.printSchema()
display(df)
```

explode()

```python
from pyspark.sql.functions import explode

df.select(
    "customer_id",
    "name",
    explode("tags").alias("tag")
).show(truncate=False)
```

explode_outer()

```python
from pyspark.sql.functions import explode_outer

df.select(
    "customer_id",
    "name",
    explode_outer("tags").alias("tag")
).show(truncate=False)
```

posexplode()

```python
from pyspark.sql.functions import posexplode

df.select(
    "customer_id",
    posexplode("tags")
).show(truncate=False)
```

posexplode_outer()


```python
from pyspark.sql.functions import posexplode_outer

df.select(
    "customer_id",
    posexplode_outer("tags").alias("position", "tag")
).show(truncate=False)
```
