```python
df = spark.table("bake_data.silver.sales_transactions_clean")
display(df)
```

```python
from pyspark.sql import functions as F

gold_df = (
    df.groupBy("product")
      .agg(
          F.sum("totalPrice").alias("total_sales"),
          F.sum("quantity").alias("total_quantity"),
          F.count("*").alias("total_transactions")
      )
)
```

```python
spark.catalog.currentCatalog()
spark.catalog.currentDatabase()
```

```python
gold_df.write \
    .mode("overwrite") \
    .format("delta") \
    .saveAsTable("bake_data.gold.product_sales")
```
