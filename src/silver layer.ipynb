```python
from pyspark.sql.functions import *

df = spark.table("bake_data.silver.sales_transactions")
```

```python
clean_df = (
    df
    .dropDuplicates()
    .fillna({"paymentMethod": "Unknown"})
    .withColumn("processed_date", current_timestamp())
)
```

```python
clean_df.write \
    .mode("overwrite") \
    .format("delta") \
    .saveAsTable("bake_data.silver.sales_transactions_clean")
```
