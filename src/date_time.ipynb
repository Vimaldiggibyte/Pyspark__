```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.getOrCreate()
```

```python
data = [
    (1, "2023-01-15", "2023-01-15 09:30:00"),
    (2, "2023-03-22", "2023-03-22 14:45:30"),
    (3, "2023-06-10", "2023-06-10 23:59:59"),
    (4, "2023-11-05", "2023-11-05 00:00:01"),
    (5, "2024-02-29", "2024-02-29 12:00:00"),  # leap year
]
```

```python
columns = ["id", "date_str", "timestamp_str"]
df = spark.createDataFrame(data, columns)
```

```python
df = df.withColumn("order_date", to_date(col("date_str"), "yyyy-MM-dd")) \
       .withColumn("order_ts", to_timestamp(col("timestamp_str"), "yyyy-MM-dd HH:mm:ss"))

df.show(truncate=False)
df.printSchema()
```

```python
from pyspark.sql.functions import to_date, to_timestamp, col, year, month, dayofmonth, dayofweek, dayofyear, weekofyear, quarter, date_format

data = [
    (1, "2023-01-15", "2023-01-15 09:30:00"),
    (2, "2023-03-22", "2023-03-22 14:45:30"),
    (3, "2023-06-10", "2023-06-10 23:59:59"),
    (4, "2023-11-05", "2023-11-05 00:00:01"),
    (5, "2024-02-29", "2024-02-29 12:00:00"),
]
columns = ["id", "date_str", "timestamp_str"]
df = spark.createDataFrame(data, columns)

df = df.withColumn("order_date", to_date(col("date_str"), "yyyy-MM-dd")) \
       .withColumn("order_ts", to_timestamp(col("timestamp_str"), "yyyy-MM-dd HH:mm:ss"))

df3 = df.select(
    "order_date",
    year("order_date").alias("year"),
    month("order_date").alias("month"),
    dayofmonth("order_date").alias("day"),
    dayofweek("order_date").alias("day_of_week"),
    dayofyear("order_date").alias("day_of_year"),
    weekofyear("order_date").alias("week_of_year"),
    quarter("order_date").alias("quarter"),
    date_format("order_date", "EEEE").alias("day_name"),
    date_format("order_date", "MMM").alias("month_short_name")
)
df3.show(truncate=False)
```

```python
df3 = df.select(
    "order_date",
    year("order_date").alias("year"),
    month("order_date").alias("month"),
    dayofmonth("order_date").alias("day"),
    dayofweek("order_date").alias("day_of_week"),   # 1=Sunday
    dayofyear("order_date").alias("day_of_year"),
    weekofyear("order_date").alias("week_of_year"),
    quarter("order_date").alias("quarter"),
    date_format("order_date", "EEEE").alias("day_name"),
    date_format("order_date", "MMM").alias("month_short_name")
)
df3.show(truncate=False)
```

```python
df6 = df.withColumn("today", current_date())
df6 = df6.withColumn("days_since_order", datediff("today", "order_date")) \
         .withColumn("months_since_order", months_between("today", "order_date"))
df6.select("order_date", "today", "days_since_order", "months_since_order").show(truncate=False)
```

```python
df7 = df.select(
    "order_date",
    trunc("order_date", "MM").alias("start_of_month"),
    trunc("order_date", "YY").alias("start_of_year"),
    date_trunc("week", "order_ts").alias("start_of_week_ts"),
    date_trunc("hour", "order_ts").alias("truncated_to_hour")
)
df7.show(truncate=False)
```

```python
df8 = df.select(
    "order_date",
    date_format("order_date", "dd/MM/yyyy").alias("uk_format"),
    date_format("order_date", "MM-dd-yyyy").alias("us_format"),
    date_format("order_ts", "yyyy-MM-dd HH:mm:ss").alias("full_ts_format")
)
df8.show(truncate=False)
```
